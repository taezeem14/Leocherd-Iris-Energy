# Leocherd Iris Energy - Kaggle Experiment 06: Model Benchmarking Suite
Measures FPS, memory footprint, and latency across batch sizes and resolutions.


```python
import time
import numpy as np

def benchmark_inference(session, input_name, shape=(1, 256, 256, 3), iterations=100):
    dummy = np.random.rand(*shape).astype(np.float32)
    # Warmup
    for _ in range(5):
        session.run(None, {input_name: dummy})
    start = time.perf_counter()
    for _ in range(iterations):
        session.run(None, {input_name: dummy})
    elapsed = time.perf_counter() - start
    fps = iterations / elapsed
    ms = (elapsed / iterations) * 1000.0
    print(f"Latency: {ms:.2f} ms | Throughput: {fps:.1f} FPS")
    return {"latency_ms": ms, "fps": fps}
```
